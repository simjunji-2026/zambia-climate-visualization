```python
import json, random
random.seed(42)

provinces_info = [
    ("Lusaka", "Lusaka", -15.41, 28.32),
    ("Copperbelt", "Ndola", -12.97, 28.63),
    ("Central", "Kabwe", -14.44, 28.45),
     ("Eastern", "Chipata", -13.64, 32.65),
     ("Western", "Mongu", -15.25, 23.13),
     ("Northern", "Kasama", -10.21, 31.18),
     ("Lupula", "Mansa", -11.18, 29.06),
     ("Northern-Western", "Solwezi", -12.18, 26.40),
     ("Southern", "Choma", -16.81, 26.99),
     ("Muchinga", "Chinsali", -10.55, 32.07),
      ]
provinces = [{"province": p, "capital": c, "latitude": la, "longitude": lo} for p,c,la,lo in provinces_info]

monthly = []
for prove, cap, la, lo in provinces_info:
    for year in range(2021, 2026):
        for month in range(1, 13):
            monthly.append({
                "data": f"{year}-{month:02d}-01",
                "province": prove,
                "capital": cap,
                "rainfall_mm": round(random.uniform(0, 220), 1),
                "tem_avg": round(random.uniform(18, 28), 1),
                "tem_max": round(random.uniform(25, 35), 1),
                "tem_min": round(random.uniform(10, 20), 1),
                "humidity_percent": round(random.uniform(40, 90),1),
                "wind_speed": round(random.uniform(2, 8), 1),
                "sunshine_hours": round(random.uniform(5, 10), 1),
                "air_quality_index": round(random.uniform(20, 120), 1),
                "pm25": round(random.uniform(5, 45), 1),
                "population": random.randint(800000, 3000000),
                "latitude": la
            })
annual = []
for prove, cap, la, lo in provinces_info:
    for year in range(2021, 2026):
        annual.append({
            "year": year,
            "province": prove,
            "annual_rainfall": round(random.uniform(600, 1200), 1),
            "avg_temp": round(random.uniform(20, 26), 1),
            "population": random.randint(800000, 3000000),
            "capital": cap,
            "latitude": la
        })
data_new = {
    "metadata": {"title": "Synthetic zambia climate, population and air-quality dataset", "period": "2021-01 to 2025-12"},
    "provinces": provinces,
    "monthly": monthly,
    "annual": annual
}
with open("zambia_synthetic_climate_2021_2025.json", "w") as f:
    json.dump(data_new, f)
print("New file saved!")
print(f"monthly: {len(monthly)}, annual: {len(annual)}, provinces: {len(provinces)}")

                
```

```python
import json
import pandas as pd

with open("zambia_synthetic_climate_2021_2025.json") as f:
    data = json.load(f)
    print(data["metadata"]["title"])
    print("Period:", data["metadata"]["period"])
    print("Sections:", list(data.keys()))

    provinces = pd.DataFrame(data["provinces"])
    monthly = pd.DataFrame(data["monthly"])
    annual = pd.DataFrame(data["annual"])

    monthly["data"] =pd.to_datetime(monthly["data"])

    print(f"monthly: {monthly.shape}, annual: {annual.shape}, provinces: {provinces.shape}")
    monthly.head()
```

```python
import matplotlib.pyplot as plt

# Make sure we have a date column - if you already did the last small cell, this does nothing bad
if "date" not in monthly.columns:
    monthly["date"] = monthly["data"]

lusaka = monthly[monthly["province"] == "Lusaka"]

fig, ax = plt.subplots()
ax.plot(lusaka["date"], lusaka["rainfall_mm"], color="blue")
ax.set_title("Lusaka: monthly rainfall, 2021-2025")
ax.set_ylabel("Rainfall (mm per monthly)")
ax.set_ylim(bottom=0)
plt.show()
```

```python
monthly["ym"] = monthly["date"].dt.to_period("M")
monthly_agg = monthly.groupby("ym").mean(numeric_only=True)
monthly_agg["rainfall_mm"] = monthly.groupby("ym")["rainfall_mm"].sum()
monthly_agg = monthly_agg.reset_index()
print("rows:", len(monthly_agg))


fig, ax = plt.subplots()
dry = monthly_agg.query("month >= 5 and month <= 10")
rainy = monthly_agg.query("month >= 11 or month <= 4")
ax.scatter(dry["rainfall_mm"], dry["pm25"], s=60, color=ORANGE, alpha=0.8, edgecolor="none", label="Dry (May-Oct)")
ax.scatter(rainy["rainfall_mm"], rainy["pm25"], s=60, color=BLUE, alpha=0.8, edgecolor="none", label="Rainy (Nov-Apr)")
ax.set_xlabel("Monthly rainfall (mm)")
ax.set_ylabel("PM2.5 (ug/m^3)")
ax.set_title("Wetter months have cleaner air")
ax.grid(axis="x", visible=True)
ax.legend(loc="upper right", markerscale=1.5)
plt.show()
r = monthly_agg["rainfall_mm"].corr(monthly_agg["pm25"])
print(f"Correlation between rainfall and PM2.5: r = {r:.2f}")
```

```python
# Get year from the date column
monthly["year"] = pd.to_datetime(monthly.data).dt.year

# Get population for each province for each year (take the mean)
pop = monthly.groupby(["province", "year"])["population"].mean().unstack()

# calculate growth % from 2021 to 2025
pop["growth"] = (pop[2025] - pop[2021]) / pop[2021] * 100

# Sort so fastest growth is at the top
pop_sorted = pop["growth"].sort_values()

# Plot horizontal bar chart
plt.figure(figsize=(10, 6))
pop_sorted.plot(kind="barh")
plt.title("Population Growth 2021 to 2025 (%)")
plt.xlabel("Growth (%)")
plt.tight_layout()
plt.show()

print(pop_sorted)
```
